# GPUPerf-3B: Phase 1 — Public Dataset Ingestion & Trajectory Mining

This notebook downloads and processes public GPU kernel benchmarks into standardized SFT and DPO training formats:
- **KernelBenchX** (`BonnieWang/KernelBenchX`): 176 tasks across 15 categories with LLM generation iterations and speedups.
- **KernelBench Mega-Traces** (`Infatoshi/kernelbench-mega-traces`): Raw agent sessions optimizing low-level GPU kernels.
- **KernelSight** (`williamhtan/kernelsight`): Hardware counter trace snapshots.

```python
# 2. Clone or import GPUPerf-3B
import os
import json
from datasets import load_dataset

os.makedirs("./data/raw", exist_ok=True)
print("Ready to download public datasets.")
```

```python
# 3. Download KernelBenchX (BonnieWang/KernelBenchX)
print("Downloading KernelBenchX...")
kbx = load_dataset("BonnieWang/KernelBenchX", "corpus_all")
print("KernelBenchX dataset structure:")
print(kbx)

kbx_path = "./data/raw/kbx_raw.jsonl"
kbx["train"].to_json(kbx_path)
print(f"Saved {len(kbx['train'])} records to {kbx_path}")
```

```python
# 4. Inspect sample record
with open(kbx_path, "r", encoding="utf-8") as f:
    sample = json.loads(f.readline())
print("Sample keys:", list(sample.keys()))
print("Task:", sample.get("task_id") or sample.get("file"))
```

```python
# 5. Download KernelBench Mega Traces (Infatoshi/kernelbench-mega-traces)
print("Downloading Mega Traces...")
try:
    mega = load_dataset("Infatoshi/kernelbench-mega-traces")
    print(mega)
    mega_path = "./data/raw/mega_traces.jsonl"
    mega["train"].to_json(mega_path)
    print(f"Saved Mega Traces to {mega_path}")
except Exception as e:
    print(f"Error loading mega traces: {e}")
```
